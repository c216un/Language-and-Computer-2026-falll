
## Part B. Response Generation

```python
def respond_emotion(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if re.search(r'해$', match.group('emotion')):
            return f'{match.group('time')} {match.group('emotion')[:-1]}하시군요. 조금 더 이야기해 주세요.'
        elif re.search(r'워$', match.group('emotion')):
            return f'{match.group('time')} {match.group('emotion')[:-1]}우시군요. 조금 더 이야기해 주세요.'
        else:
            return f'{match.group('time')} {match.group('emotion')[:-1]}으시군요. 조금 더 이야기해 주세요.'
    return None

def respond_desire(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        return f'{match.group('desire')} 싶으시군요. 조금 더 이야기해 주세요.'
    return None


def respond_cause(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if ((ord(match.group('cause')[-1])) - 44032) % 28 == 0:
            return f'{match.group('cause')}가 {match.group('status')}서 신경쓰이시군요. 조금 더 이야기해 주세요.'
        else:
            return f'{match.group('cause')}이 {match.group('status')}서 신경쓰이시군요. 조금 더 이야기해 주세요.'

    return None



def respond_relationship(user_input, pattern):
    match = re.search(pattern, user_input)


    if match:
        behavior = match.group('behavior')

         # '나를', '나에게'와 주변 공백을 정리
        behavior = re.sub(r'\s*나(?:를|에게)\s*', ' ', behavior).strip()

        if ((ord(match.group('relationship')[-2])) - 44032) % 28 == 0: # 종성 유무 확인
            return f'{match.group('relationship')[:-1]}가 {behavior}서 신경쓰이시군요. 조금 더 이야기해 주세요.'
        else:
            return f'{match.group('relationship')[:-1]}이 {behavior}서 신경쓰이시군요. 조금 더 이야기해 주세요.'

    return None


def respond_preference(user_input, pattern):
    match = re.search(pattern, user_input)
    
    if match:
        if ((ord(match.group('object')[-2])) - 44032) % 28 == 0 and re.search(r'좋', match.group('preference')):
            return f'{match.group('object')[:-1]}가 좋으시군요. 조금 더 이야기해 주세요.'
        elif ((ord(match.group('object')[-2])) - 44032) % 28 == 0 and re.search(r'싫', match.group('preference')):
            return f'{match.group('object')[:-1]}가 싫으시군요. 조금 더 이야기해 주세요.'
        elif ((ord(match.group('object')[-2])) - 44032) % 28 != 0 and re.search(r'좋', match.group('preference')):
                    return f'{match.group('object')[:-1]}이 좋으시군요. 조금 더 이야기해 주세요.'
        elif ((ord(match.group('object')[-2])) - 44032) % 28 != 0 and re.search(r'싫', match.group('preference')):
                return f'{match.group('object')[:-1]}이 싫으시군요. 조금 더 이야기해 주세요.'
        else: 
            return None

    return None


def respond_question(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if re.search(r'해$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}하는지 궁금하시군요. 잠시만 기다려 주세요.'
        elif re.search(r'어$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}는지 궁금하시군요. 잠시만 기다려 주세요.'
        elif re.search(r'나$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}나는지 궁금하시군요. 잠시만 기다려 주세요.'
        elif re.search(r'지$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}은지 궁금하시군요. 잠시만 기다려 주세요.'
        else:
            return None
        
    return None


RESPONSE_FUNCTIONS = {
    "emotion": respond_emotion,
    "desire": respond_desire,
    "cause": respond_cause,
    "relationship": respond_relationship,
    "preference": respond_preference,
    "question": respond_question,
}
```

## Part C. Chatbot Implementation

```python
def respond(user_input):
    for key, pattern in PATTERNS.items():
        if re.search(pattern, user_input):
            return RESPONSE_FUNCTIONS[key](user_input, pattern)
    return "잘 이해하지 못했어요." 
```

- 더 적절하다고 생각하는 response: 시험 때문에 너무 피곤하시군요. 조금 더 이야기해 주세요.

4. 우리 친구는 나를 자꾸 무시해

- 처리하지 못한 이유: relationship 정규식은 ‘내 + 관계’ 형태만 허용하고 ‘우리 + 관계’ 형태는 허용하지 않는다.

5. 나는 고양이가 마음에 들어

- 더 적절하다고 생각하는 response: 누가 파이썬을 개발했는지 궁금하시군요. 잠시만 기다려 주세요.
- 처리하지 못한 이유: 질문 정규식은 문장에 매칭되지만, 응답 함수가 문장 끝의 ‘지’를 일괄적으로 ‘은지’로 바꾼다. ‘개발했지’는 ‘개발했는지’로 바뀌어야 하므로 현재 변환 규칙으로는 적절한 응답을 만들지 못한다.

#### 2. 위 오류들 각각에 대하여, 해당 오류를 줄이기 위해 regular expression을 어떻게 수정할 수 있을지 제안하세요. 실제 코드를 수정할 필요는 없습니다.

1. 시간 표현과 ‘나는’의 두 가지 순서를 모두 인식하도록 수정한다. ‘시간 표현 + 나는’뿐 아니라 ‘나는 + 시간 표현’도 허용한다.

정규표현식 제안:
r"(?:(?P<time_before>요즘|오늘|가끔)\s나는|나는\s(?P<time_after>요즘|오늘|가끔))\s(?P<degree>너무|좀|많이|꽤|이상하게)?\s?(?P<emotion>.+)"

응답 함수에서는 time_before와 time_after 중 값이 있는 그룹을 사용해야 한다.

2. emotion 정규식의 시간 표현 목록에 ‘오늘따라’를 추가한다. 더 긴 표현이 먼저 매칭되도록 ‘오늘따라’를 ‘오늘’보다 앞에 둔다.

정규표현식 제안:
r"(?P<time>오늘따라|요즘|오늘|가끔)\s나는\s(?P<degree>너무|좀|많이|꽤|이상하게)?\s?(?P<emotion>.+)"

또한 preference 정규식이 ‘대상 + 조사 + 선호 표현’ 형태를 요구하도록 수정하면 ‘기분이 안 좋아’가 선호 표현으로 잘못 매칭되는 경우를 줄일 수 있다.

정규표현식 제안:
r"나는\s(?P<object>.+?)(?P<particle>을|를|이|가|도)\s(?P<preference>좋아해|좋아|싫어)"

3. ‘때문에’를 별도 그룹으로 잡도록 수정할 수 있다.

정규표현식 제안:
r"(?P<cause>.+?)\s(?P<cause_marker>때문에)\s(?P<status>.+)"

이 정규식은 원인과 상태를 구분해 포착한다. 다만 자연스러운 응답을 만들려면 응답 함수도 cause_marker의 ‘때문에’를 살려 문장을 구성해야 한다.

4. 관계 앞의 표현으로 ‘내’와 ‘우리’를 모두 허용하고, 관계 표현과 주격 조사를 나누어 포착한다.

정규표현식 제안:
r"(?:내|우리)\s(?P<relationship>\w+?)(?P<subject_particle>은|는|이|가)\s(?P<behavior>.+)"

5. 선호 표현에 ‘마음에 들어’를 추가하고, 대상과 조사를 나누어 포착한다.

정규표현식 제안:
r"나는\s(?P<object>.+?)(?P<particle>을|를|이|가|도)\s(?P<preference>좋아해|좋아|싫어|마음에 들어)"

이 표현에 알맞은 응답을 만들려면 응답 함수에도 ‘마음에 드시는군요’ 형태의 처리 규칙이 필요하다.

6. 질문 문장 끝의 표현을 별도 그룹으로 포착하고, ‘했지’를 다른 끝 표현과 구분한다.

정규표현식 제안:
r"(?P<question_word>누가|언제|어디서|무엇을|어떻게|왜)\s(?P<question_content>.+?)(?P<ending>했지|해|어|나|지)?"

응답 함수에서 ending이 ‘했지’인 경우에는 ‘했는지’로 바꾸도록 처리해야 한다. ‘지’를 항상 ‘은지’로 바꾸면 ‘개발했지’와 같은 문장을 올바르게 변환할 수 없다.
