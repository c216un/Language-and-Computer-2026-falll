#### emotion
빈도/시간: 요즘, 오늘, 가끔  
\+  
주어; 나는  
\+  
정도부사: 너무, 좀, 많이, 꽤, 이상하게(?) -> 있거나 없거나  
\+  
서술어 패턴:  
    1) OO해; 피곤해, 불안해, 답답해, 우울해, 예민해  -> 해로 끝남 -> 하시군요  
    2) 기분이 안 좋아  -> (기분이 안 좋)으시군요 else -> 으시군요  
    3) 외로워  -> (외로)우시군요; 안쓰러워, 고마워 등... -> 워로 끝남 -> 우시군요  
    4) 지쳐 있어  -> (지쳐 있)으시군요 else -> 으시군요  


(?p\<time>\w{2}) (\w{2}) (?p\<degree>\w*) (?p\<emotion>.+)   
(?p\<emotion>.+) 받아서 또 (\w$) # 응답 서술어 어간(-으시, -우시) 판단

#### desire
주어; 나는  
+  
집에/친구를/그냥/맛있는 걸/ 새로운 취미를/주말에는 아무것도 안  
+  
-고  
+  
싶어 or 싶다  

(\w{2}) (?p\<desire>.+고$) (\w{2})

#### relationship

#### preference
나는  
+  
선호 대상  
+  
을/를/이/가  
+  
좋아, 좋아해, 싫어  

(\w{2}) (?p\<object>.+) (?p\<preference>.+) 

#### question
의문사(왜, 어떻게, 언제, 어디서)  
+  
궁금한 내용  

(?p\<question_word>\S+) (?p\<question_content>.+)?

응답: 의문사 + 내용 + -지 궁금하시군요. 

피곤해 -> 피곤하신지  
거짓말을 해 -> 거짓말을 하는지  
잘 수 있어 -> 잘 수 있는지  
끝나 -> 끝나는지  
살 수 있어 -> 살 수 있는지  
하기 싫지 -> 하기 싫은지  
잘할 수 있어 -> 잘할 수 있는지  
시작해 -> 시작하는지   

-해 -> -하는지   
-어 -> -는지 % 싫어? -> 싫은지 (왜???)  
-나 -> -나는지  
-지 -> -은지
