## Part A. Pattern Design

```python
# json 파일 읽어 DATA dict에 저장
import json

with open('data.json','r') as f:
    DATA = json.load(f)
    print(DATA)
```

```python
# regular-expression patterns 작성 및 dict에 저장

PATTERNS = {
    "emotion": r"(?P<time>\w{2})\s나는\s(?P<degree>너무|좀|많이|꽤|이상하게)?\s?(?P<emotion>.+)",
    "desire": r"나는\s(?P<desire>.+고)\s(싶어|싶다)", 
    "cause": r"(?P<cause>\w+)\s때문에\s(?P<status>.+)",
    "relationship": r"내\s(?P<relationship>\w+)(은|는|이|가)\s(?P<behavior>.+)",
    "preference": r"나는\s(?P<object>.+)(을|를|이|가)\s(?P<preference>좋아해|좋아|싫어)",
    "question": r"(?P<question_word>왜|어떻게|언제|어디서)\s(?P<question_content>.+)\?", 
}
```

#### emotion
빈도/시간: 요즘, 오늘, 가끔  
\+  
주어; 나는  
\+  
정도부사: 너무, 좀, 많이, 꽤, 이상하게(?) -> 있거나 없거나  
\+  
서술어 패턴:  
    1) OO해; 피곤해, 불안해, 답답해, 우울해, 예민해  -> 해로 끝남 -> 하시군요  
    2) 기분이 안 좋아  -> (기분이 안 좋)으시군요 else -> 으시군요  
    3) 외로워  -> (외로)우시군요; 안쓰러워, 고마워 등... -> 워로 끝남 -> 우시군요  
    4) 지쳐 있어  -> (지쳐 있)으시군요 else -> 으시군요  


(?P\<time>\w{2})\s나는\s(?P\<degree>\w*)\s?(?P\<emotion>.+)   
(?P\<emotion>.+) 받아서 또 (\w$) # 응답 서술어 어간(-으시, -우시) 판단

```python
import re

emotion_input = DATA['emotion']
processed_input = []
groups_input = []

for i in range(len(emotion_input)):
    m = re.search(r'(?P<time>\w{2})\s나는\s(?P<degree>너무|좀|많이|꽤|이상하게)?\s?(?P<emotion>.+)',emotion_input[i])
    processed_input.append(m)
    groups_input = [m.group(0), m.group('time'), m.group('degree'), m.group('emotion')]
    print(groups_input)
                                   
```

#### desire
주어; 나는  
+  
집에/친구를/그냥/맛있는 걸/ 새로운 취미를/주말에는 아무것도 안  
+  
-고  
+  
싶어 or 싶다  

나는\s(?P\<desire>.+고)\s(싶어|싶다)



```python
import re

desire_input = DATA['desire']
processed_input = []
groups_input = []

for i in range(len(desire_input)):
    m = re.search(r'나는\s(?P<desire>.+고)\s(싶어|싶다)',desire_input[i])
    processed_input.append(m)
    groups_input = [m.group(0), m.group('desire')]
    print(groups_input)
```

#### cause
이유  
+  
때문에  
+  
상태  

(?P\<cause>\w+)\s때문에\s(?P\<status>.+)

% 응답 시 상태를 어떻게 표현할지? 

```python
import re

cause_input = DATA['cause']
processed_input = []
groups_input = []

for i in range(len(cause_input)):
    m = re.search(r'(?P<cause>\w+)\s때문에\s(?P<status>.+)',cause_input[i])
    processed_input.append(m)
    groups_input = [m.group(0), m.group('cause'), m.group('status')]
    print(groups_input)
```

#### relationship
내  
+  
관계  
+  
은/는/이/가  
+  
행동

내\s(?P\<relationship>\w+)(은|는|이|가)\s(?P\<behavior>.+) 

응답: ~이|가(받침에 따라) ~서 신경쓰이시군요

```python
import re

relationship_input = DATA['relationship']
processed_input = []
groups_input = []

for i in range(len(relationship_input)):
    m = re.search(r'내\s(?P<relationship>\w+)(은|는|이|가)\s(?P<behavior>.+)',relationship_input[i])
    processed_input.append(m)
    groups_input = [m.group(0), m.group('relationship'), m.group('behavior')]
    print(groups_input)
```

#### preference
나는  
+  
선호 대상  
+  
을/를/이/가  
+  
좋아, 좋아해, 싫어  

나는\s(?P\<object>.+)(을|를|이|가)\s(?P\<preference>좋아|좋아해|싫어) 

```python
import re

preference_input = DATA['preference']
processed_input = []
groups_input = []

for i in range(len(preference_input)):
    m = re.search(r'나는\s(?P<object>.+)(을|를|이|가)\s(?P<preference>좋아해|좋아|싫어)',preference_input[i])
    processed_input.append(m)
    groups_input = [m.group(0), m.group('object'), m.group('preference')]
    print(groups_input)
```

#### question
의문사(왜, 어떻게, 언제, 어디서)  
+  
궁금한 내용  

(?P\<question_word>왜|어떻게|언제|어디서)\s(?P\<question_content>.+)\\?

응답: 의문사 + 내용 + -지 궁금하시군요. 

피곤해 -> 피곤하신지  
거짓말을 해 -> 거짓말을 하는지  
잘 수 있어 -> 잘 수 있는지  
끝나 -> 끝나는지  
살 수 있어 -> 살 수 있는지  
하기 싫지 -> 하기 싫은지  
잘할 수 있어 -> 잘할 수 있는지  
시작해 -> 시작하는지   

-해 -> -하는지   
-어 -> -는지 % 싫어? -> 싫은지 (왜???)  
-나 -> -나는지  
-지 -> -은지  

```python
import re

question_input = DATA['question']
processed_input = []
groups_input = []

for i in range(len(question_input)):
    m = re.search(r'(?P<question_word>왜|어떻게|언제|어디서)\s(?P<question_content>.+)\?',question_input[i])
    processed_input.append(m)
    groups_input = [m.group(0), m.group('question_word'), m.group('question_content')]
    print(groups_input)
```


## Part B. Response Generation

```python
def respond_emotion(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if re.search(r'해$', match.group('emotion')):
            return f'{match.group('time')} {match.group('emotion')[:-1]}하시군요. 조금 더 이야기해 주세요.'
        elif re.search(r'워$', match.group('emotion')):
            return f'{match.group('time')} {match.group('emotion')[:-1]}우시군요. 조금 더 이야기해 주세요.'
        else:
            return f'{match.group('time')} {match.group('emotion')[:-1]}으시군요. 조금 더 이야기해 주세요.'
    return None

def respond_desire(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        return f'{match.group('desire')} 싶으시군요. 조금 더 이야기해 주세요.'
    return None


def respond_cause(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if ((ord(match.group('cause')[-1])) - 44032) % 28 == 0:
            return f'{match.group('cause')}가 {match.group('status')}서 신경쓰이시군요. 조금 더 이야기해 주세요.'
        else:
            return f'{match.group('cause')}이 {match.group('status')}서 신경쓰이시군요. 조금 더 이야기해 주세요.'

    return None



def respond_relationship(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if ((ord(match.group('relationship')[-1])) - 44032) % 28 == 0: # 종성 유무 확인
            return f'{match.group('relationship')}가 {match.group('behavior')}서 신경쓰이시군요. 조금 더 이야기해 주세요.'
        else:
            return f'{match.group('relationship')}이 {match.group('behavior')}서 신경쓰이시군요. 조금 더 이야기해 주세요.'

    return None


def respond_preference(user_input, pattern):
    match = re.search(pattern, user_input)
    
    if match:
        if ((ord(match.group('object')[-1])) - 44032) % 28 == 0 and re.search(r'좋', match.group('preference')):
            return f'{match.group('object')}가 좋으시군요. 조금 더 이야기해 주세요.'
        elif ((ord(match.group('object')[-1])) - 44032) % 28 == 0 and re.search(r'싫', match.group('preference')):
            return f'{match.group('object')}가 싫으시군요. 조금 더 이야기해 주세요.'
        elif ((ord(match.group('object')[-1])) - 44032) % 28 != 0 and re.search(r'좋', match.group('preference')):
                    return f'{match.group('object')}이 좋으시군요. 조금 더 이야기해 주세요.'
        elif ((ord(match.group('object')[-1])) - 44032) % 28 != 0 and re.search(r'싫', match.group('preference')):
                return f'{match.group('object')}이 싫으시군요. 조금 더 이야기해 주세요.'
        else: 
            return None

    return None


def respond_question(user_input, pattern):
    match = re.search(pattern, user_input)

    if match:
        if re.search(r'해$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}하는지 궁금하시군요. 잠시만 기다려 주세요.'
        elif re.search(r'어$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}는지 궁금하시군요. 잠시만 기다려 주세요.'
        elif re.search(r'나$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}나는지 궁금하시군요. 잠시만 기다려 주세요.'
        elif re.search(r'지$', match.group('question_content')):
            return f'{match.group('question_word')} {match.group('question_content')[:-1]}은지 궁금하시군요. 잠시만 기다려 주세요.'
        else:
            return None
        
    return None


RESPONSE_FUNCTIONS = {
    "emotion": respond_emotion,
    "desire": respond_desire,
    "cause": respond_cause,
    "relationship": respond_relationship,
    "preference": respond_preference,
    "question": respond_question,
}
```

## Part C. Chatbot Implementation

```python
def respond(user_input):
    for key, pattern in PATTERNS.items():
        if re.search(pattern, user_input):
            return RESPONSE_FUNCTIONS[key](user_input, pattern)
    return "잘 이해하지 못했어요." 
```

## Part D. Demo & Interactive Chat

```python
demo_inputs = [
    "요즘 나는 피곤해",
    "나는 친구를 만나고 싶어",
    "과제 때문에 너무 바빠",
    "내 형이 자꾸 나를 놀려",
    "나는 조용한 카페를 좋아해",
    "언제 방학이 시작해?",
]
```

```python
for text in demo_inputs:
    print("USER:", text)
    print("BOT :", respond(text))
    print()
```

```python
# Interactive Chat
# '종료'를 입력하면 대화를 끝낼 수 있습니다.
# Type '종료' to end the interactive chat.

while True:
    user_input = input("YOU: ")

    if user_input.strip() == "종료":
        print("BOT: 대화를 종료합니다.")
        break

    print("BOT:", respond(user_input))
    print()
```
