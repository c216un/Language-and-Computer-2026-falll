```python
f = open('textfile.txt', 'r')
content = f.read()
print(content)
f.close()
```

```python
with open('textfile.txt', 'r') as f:
    content = f.read()
print(content)
```

```python
with open('textmultilines.txt', 'r') as f:
    lines = f.readlines()
    for line in lines:
        print(line)
```
