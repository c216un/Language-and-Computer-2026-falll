## Table Data
#### CSV(TSV)

```python
newlist = []

with open('sample_data.csv', 'r') as f:
    lines = f.readlines()
    for line in lines:
        info = line.strip().split(',')
        if info[0] == 'id':
            continue
        newlist.append(info)

for line in newlist:
    print(line)
```

```python
import csv # Python 내장 라이브러리

# csv 파일 열기
with open('sample_data.tsv', 'r') as f:
    # csv.reader: csv 파일의 내용을 읽어들여 레코드와 필드로 나누어 줌
    reader = csv.reader(f, delimiter='\t')

    print(reader)
    print(list(reader))
    
```

```python
with open('sample_data.tsv', 'r') as f:
    reader = csv.reader(f, delimiter='\t')

    for record in reader:
        print(record)
```

```python
with open('sample_data.tsv', 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader) # 첫번째 행을 건너뛰고 다음 행부터 접근

    for record in reader:
        print(record)
```

```python
# 잘못된 구분자를 사용하면 필드를 올바르게 나눌 수 없다.

with open('sample_data.tsv', 'r') as f:
    reader = csv.reader(f, delimiter=',')
    next(reader)

    for record in reader:
        print(record)
```

```python
with open('sample_data.csv', 'r') as f:
    reader = csv.reader(f, delimiter='\t')
    next(reader)

    for record in reader:
        print(record)
```

```python
with open('sample_new.csv', 'w') as f:
    writer = csv.writer(f, delimiter=',') # csv.writer 설정: 파일에 쓰기 준비
    writer.writerow(['id', 'name', 'age']) # 레코드 하나씩을 파일에 쓴다
    writer.writerow(['1', 'Alice', '25']) # 리스트의 원소 하나가 필드 하나에 대응
    writer.writerow(['2', 'Bob', '30'])
    writer.writerow(['3', 'Candy', '27'])

!cat sample_new.csv
```

#### Object File: pickle

```python
import random
import pickle
```

```python
randlist = random.sample(range(1, 10000), 1000)
# 길이 1000인 리스트
print(randlist)
```

```python
with open('randlist.pkl', 'wb') as f:
    pickle.dump(randlist, f)
```

```python
with open('randlist.pkl', 'rb') as f:
    loaded = pickle.load(f)

print(loaded)
```
