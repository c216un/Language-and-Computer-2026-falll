## Regular Expression
#### Basic Expressions

```python
import re

print(re.findall('서울대입구', '서울대 서울대입구 서울대입구역 북서울대입구'))
print(re.findall('.서울대입구역?', '서울대 서울대입구 서울대입구역 북서울대입구'))
```

#### Disjunctions []

```python
print(re.findall('[1234567890]', 'plenty of 7 to 5'))
```

```python
print(re.findall('[1234567890]', '1925 to 1983'))
```

```python
print(re.findall('[0-9]', '1925 to 1983'))
print(re.findall('[9-0]', '1925 to 1983'))
```

```python
print(re.findall('[언국]어와컴퓨터', '언어와컴퓨터 국어와컴퓨터 독어와컴퓨터'))
print(re.findall('[^언]어와컴퓨터', '언어와컴퓨터 국어와컴퓨터 독어와컴퓨터'))
print(re.findall('[^언국]어와컴퓨터', '언어와컴퓨터 국어와컴퓨터 독어와컴퓨터'))
```

#### Disjunctions |

```python
print(re.findall('Python|Java|C', 'Language: Python, Java, and C'))
```

```python
print(re.findall('언|국어와컴퓨터', '언어와컴퓨터 국어와컴퓨터 독어와컴퓨터'))
```

#### Optional Matching ?

```python
print(re.findall('설대?입구?', '서울대입구 설대입구 설입'))
```

```python
print(re.findall('[rn]?ice', 'rice vice nice ice'))
```

#### Repetition +, * (Kleene)

```python
print(re.findall('baa+!', 'b ba ba! baa! baa baaa! baaaa!'))
print(re.findall('baa*!', 'b ba ba! baa! baa baaa! baaaa!'))
```

#### Repetition {n, m}

```python
print(re.findall('ba{2,3}!', 'b ba ba! baa! baa baaa! baaaa!'))
```

#### Repetition of Multiple Characters

```python
print(re.search(r'(ba)+!', 'b!'))
print(re.search(r'(ba)+!', 'ba!'))
print(re.search(r'(ba)+!', 'baa!'))
print(re.search(r'(ba)+!', 'baba!'))
print(re.search(r'(ba)+!', 'baaa!'))
print(re.search(r'(ba)+!', 'bababa!'))
```

#### Wildcard .

```python
print(re.findall('.대입구', '대입구 설대입구 공대입구 건대입구'))
```

```python
print(re.findall('.대입구', '총신대입구'))
print(re.findall('..대입구', '총신대입구'))
print(re.findall('.*대입구', '총신대입구'))
```

#### Anchors

```python
print(re.findall('^안녕$', '안녕'))
print(re.findall('^안녕$', '안녕하세요'))
print(re.findall('^안녕$', '아안녕'))
```

#### Escape Code

```python
print(re.findall(r'\bis\b', 'This is an isalnd.'))
print(re.findall(r'\Bis', 'This is an isalnd.'))
print(re.findall(r'is\B', 'This is an isalnd.'))
```

#### Escape Code

```python
print(re.findall('\.', '안녕하세요.'))
print(re.findall('.', '안녕하세요.'))
```

```python
s = '안녕하세요\n이것은 긴 문자열입니다'
print(re.findall('\n', s))
```

```python
ss = """안녕하세요
이것은 긴 문자열입니다"""
print(re.findall('\n', ss))
```

#### Other Expressions

```python
print(re.findall('\d', '2022년 10월 8일'))
print(re.findall('\D', '2022년 10월 8일'))
print(re.findall('\w', '2022년 10월 8일'))
print(re.findall('\W', '2022년 10월 8일'))
print(re.findall('\s', '2022년 10월 8일'))
print(re.findall('\S', '2022년 10월 8일'))
```

#### Raw String r

```python
print(re.findall(r'is\b', 'This is an island.'))
print(re.findall('is\b', 'This is an island.'))
```

```python
print('hello\b')
```

#### Hierarchy

```python
print(re.search('(^|[^a-zA-Z])[tT]he([^a-zA-Z]|$)', 'the'))
print(re.search('(^|[^a-zA-Z])[tT]he([^a-zA-Z]|$)', 'The'))
print(re.search('(^|[^a-zA-Z])[tT]he([^a-zA-Z]|$)', 'they'))
print(re.search('(^|[^a-zA-Z])[tT]he([^a-zA-Z]|$)', 'either'))
print(re.search('(^|[^a-zA-Z])[tT]he([^a-zA-Z]|$)', 'the5'))
```

#### Greedy Matching

```python
print(re.findall('[a-z]+', 'hello, world!'))
```

```python
print(re.findall('[a-z]+?', 'hello, world!'))
```

```python
print(re.findall('hello[0-9]*', 'hello999world!'))
print(re.findall('hello[0-9]*?', 'hello999world!'))
```

## RE with Python
#### re Library

```python
import re
```

#### RE Object

```python
ptn = '[0-9]+'
print(type(ptn))
```

```python
ptn_obj = re.compile(ptn)
print(type(ptn_obj))
```

```python
searched = ptn_obj.search('은하철도999')
print(searched)
```

#### search: 문자열 안에서 패턴 매칭하기

```python
numbs_1 = re.search(ptn_obj, '은하철도999')
print(numbs_1)

numbs_2 = re.search('[0-9]+', '은하철도999')
print(numbs_2)
```

```python
searched = ptn_obj.search('은하철도999')
print(searched)
```

```python
print(re.search(ptn_obj, '1초라도안보이면 2렇게초조한데'))
```

```python
if numbs_1:
    print('Yes, matched')
else:
    print('Not matched')
```

```python
print(re.search(ptn_obj, '기차가어둠을헤치고'))
```

```python
train = re.search(ptn_obj,'기차가어둠을헤치고')
print(train)

if train:
    print('matched')
else:
    print('not matched')
```

#### match: 문자열 맨 앞부터 패턴 매칭하기

```python
print(re.match('P', 'Python Programming'))
print(re.match('P', 'Java Programming'))
```

```python
p_obj = re.compile('[A-Z]+')
print(p_obj.match('python Programming'))
print(p_obj.search('python Programming'))
```

#### findall: 문자열 안에서 패턴 모두 찾기

```python
ptn_obj = re.compile('[0-9]+')
print(re.findall(ptn_obj, '1초라도안보이면 2렇게초조한데'))
```

```python
matches = re.findall('[AEIOU][A-Za-z]+', 'Alice, Nina, Joe and Euqene')
print(matches)
```

#### split: 문자열을 패턴 기준으로 쪼개기

```python
print(re.split('[0-9]+','안1녕222하983세00요7!'))
```

```python
splits = re.split(r'\B', '안녕하세요 이것은 샘플 문장입니다')
print(splits)
```

#### sub: 문자열 안에서 패턴 찾아 바꾸기

```python
sample_str = '그 물건은 2022년 9월 30일 5000원에 판매되었다.'
num_ptn = re.compile('[0-9]+')

subed = re.sub(num_ptn, 'X', sample_str)
print(subed)
```

```python
sample_str = '그 물건은 2022년 9월 30일 5000원에 판매되었다.'
num_ptn_2 = re.compile('[0-9]+?')

subed_2 = re.sub(num_ptn_2, 'X', sample_str)
print(subed_2)
```
